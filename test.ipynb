```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from src.data_loader import load_data
from sklearn.linear_model import  LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
data = load_data()
```

```python
x = data.drop(["Outcome"] , axis=1)
y = data["Outcome"]
train_x,test_x,train_y,test_y = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
lg_reg = LogisticRegression()
lg_reg.fit(train_x,train_y)
```

```python
test_pred = lg_reg.predict(test_x)
accuracy_score(test_pred,test_y)
```

```python
from sklearn.metrics import classification_report

print("\nClassification Report:")
print(classification_report(test_y, test_pred))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(test_y, test_pred)

display = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["No Diabetes", "Diabetes"]
)

display.plot()
plt.title("Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
import joblib
from pathlib import Path

MODEL_PATH = Path.cwd() / "diabetes_model.pkl"

joblib.dump(lg_reg, MODEL_PATH)

print(f"Model saved to: {MODEL_PATH}")
```

```python
loaded_model = joblib.load(MODEL_PATH)

loaded_predictions = loaded_model.predict(test_x)

print("Predictions match:", (loaded_predictions == test_pred).all())
```
